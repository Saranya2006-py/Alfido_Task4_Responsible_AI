```python
df_filled = df.copy()

for col in df_filled.columns:
    if df_filled[col].dtype in ["int64", "float64"]:
        df_filled[col] = df_filled[col].fillna(df_filled[col].median())
    else:
        df_filled[col] = df_filled[col].fillna(df_filled[col].mode()[0])
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test_scaled)
accuracy_score(y_test, y_pred)
```

```python
# Get predictions on test data
y_test_pred = model.predict(X_test_scaled)

# Create a DataFrame for fairness analysis
fairness_df = X_test.copy()
fairness_df["actual"] = y_test.values
fairness_df["prediction"] = y_test_pred
```

```python
# Check available columns
[f for f in fairness_df.columns if "sex" in f]
```

```python
# Group by sex and calculate positive prediction rate
fairness_metrics = fairness_df.groupby("sex")["prediction"].mean()
fairness_metrics
```

## Fairness Analysis by Gender

The model's predictions were analyzed across gender groups using the
sex_Male feature.

- Females (sex_Male = False) have a positive prediction rate of approximately 25.7%.
- Males (sex_Male = True) have a positive prediction rate of approximately 65.1%.

This shows a significant disparity in prediction rates between genders.


## Responsible AI Discussion

The observed disparity may be influenced by real-world medical data,
